```python
# ================================================================
# PHASE 5B — CELL 4
# VERIFY FROZEN SEGMENTATION MODEL IDENTITY
# ================================================================

EXPECTED_SEGMENTATION_MODEL_SHA256 = (
    "65ec53959831c6b7c08747a5d94de656"
    "3580ce9ec052aa7b9afa614bee5705d3"
)


def calculate_sha256(file_path):

    sha256_hash = hashlib.sha256()

    with open(file_path, "rb") as file:

        for block in iter(
            lambda: file.read(1024 * 1024),
            b""
        ):
            sha256_hash.update(block)

    return sha256_hash.hexdigest()


actual_segmentation_model_sha256 = (
    calculate_sha256(
        FROZEN_SEGMENTATION_MODEL_PATH
    )
)


print("=" * 80)
print("FROZEN SEGMENTATION MODEL IDENTITY")
print("=" * 80)

print(
    "\nExpected SHA256:"
)

print(
    EXPECTED_SEGMENTATION_MODEL_SHA256
)

print(
    "\nActual SHA256:"
)

print(
    actual_segmentation_model_sha256
)


assert (
    actual_segmentation_model_sha256
    == EXPECTED_SEGMENTATION_MODEL_SHA256
), (
    "Segmentation model SHA256 mismatch. "
    "Do not continue."
)


print(
    "\n✓ Exact frozen Phase 1 segmentation model verified."
)
```

```python
# ================================================================
# PHASE 5B — CELL 7
# DISCOVER PHASE 2 AND PHASE 3 IMPLEMENTATION ARTIFACTS
# ================================================================

PHASE2_DIR = (
    PROJECT_ROOT
    / "phase2_quantitative"
)

PHASE3_DIR = (
    PROJECT_ROOT
    / "phase3_classification"
)


def list_project_files(
    root_directory,
    allowed_suffixes=None
):

    records = []

    for path in root_directory.rglob("*"):

        if not path.is_file():
            continue

        if (
            allowed_suffixes is not None
            and path.suffix.lower()
            not in allowed_suffixes
        ):
            continue

        records.append({
            "relative_path":
                str(
                    path.relative_to(
                        PROJECT_ROOT
                    )
                ),

            "suffix":
                path.suffix.lower(),

            "size_kb":
                round(
                    path.stat().st_size / 1024,
                    2
                )
        })

    return pd.DataFrame(records)


artifact_suffixes = {
    ".json",
    ".csv",
    ".pkl",
    ".pickle",
    ".joblib",
    ".keras",
    ".h5",
    ".npz",
    ".npy",
    ".txt"
}


phase2_artifacts_df = list_project_files(
    PHASE2_DIR,
    artifact_suffixes
)

phase3_artifacts_df = list_project_files(
    PHASE3_DIR,
    artifact_suffixes
)


print("=" * 100)
print("PHASE 2 SAVED ARTIFACTS")
print("=" * 100)

display(
    phase2_artifacts_df
)


print("\n" + "=" * 100)
print("PHASE 3 SAVED ARTIFACTS")
print("=" * 100)

display(
    phase3_artifacts_df
)


print(
    f"\nPhase 2 files found: "
    f"{len(phase2_artifacts_df)}"
)

print(
    f"Phase 3 files found: "
    f"{len(phase3_artifacts_df)}"
)
```

```python
# ================================================================
# PHASE 5B — CELL 16
# REPRODUCE FROZEN PHASE 3 PREDICTIONS
# ================================================================

phase3_verification_features = (
    phase3_frozen_predictions_df[
        PHASE3_FEATURE_COLUMNS
    ].copy()
)


reproduced_predictions = (
    phase3_classifier_artifact.predict(
        phase3_verification_features
    )
)


reproduced_probabilities = (
    phase3_classifier_artifact.predict_proba(
        phase3_verification_features
    )
)


assert reproduced_probabilities.shape == (
    len(phase3_frozen_predictions_df),
    3
)


phase3_classifier_verification_df = pd.DataFrame({

    "patient_id":
        phase3_frozen_predictions_df[
            "patient_id"
        ].values,

    "frozen_prediction":
        phase3_frozen_predictions_df[
            "learned_prediction"
        ].values,

    "reproduced_prediction":
        reproduced_predictions,

    "frozen_prob_Higher_EF":
        phase3_frozen_predictions_df[
            "prob_Higher_EF"
        ].values,

    "reproduced_prob_Higher_EF":
        reproduced_probabilities[:, 0],

    "frozen_prob_Intermediate_EF":
        phase3_frozen_predictions_df[
            "prob_Intermediate_EF"
        ].values,

    "reproduced_prob_Intermediate_EF":
        reproduced_probabilities[:, 1],

    "frozen_prob_Lower_EF":
        phase3_frozen_predictions_df[
            "prob_Lower_EF"
        ].values,

    "reproduced_prob_Lower_EF":
        reproduced_probabilities[:, 2]
})


phase3_classifier_verification_df[
    "prediction_match"
] = (
    phase3_classifier_verification_df[
        "frozen_prediction"
    ]
    ==
    phase3_classifier_verification_df[
        "reproduced_prediction"
    ]
)


print("=" * 90)
print("PHASE 3 CLASSIFIER REPRODUCTION CHECK")
print("=" * 90)

print(
    "\nPrediction matches:",
    phase3_classifier_verification_df[
        "prediction_match"
    ].sum(),
    "/",
    len(
        phase3_classifier_verification_df
    )
)


display(
    phase3_classifier_verification_df.head()
)
```

```python
# ================================================================
# PHASE 5B — CELL 17
# VERIFY NUMERICAL PROBABILITY REPRODUCTION
# ================================================================

probability_checks = {}


for class_name, probability_index in [

    ("Higher_EF", 0),

    ("Intermediate_EF", 1),

    ("Lower_EF", 2)

]:

    frozen_values = (
        phase3_frozen_predictions_df[
            f"prob_{class_name}"
        ].to_numpy()
    )

    reproduced_values = (
        reproduced_probabilities[
            :,
            probability_index
        ]
    )

    absolute_difference = np.abs(
        frozen_values
        -
        reproduced_values
    )

    probability_checks[
        class_name
    ] = {

        "max_absolute_difference":
            float(
                absolute_difference.max()
            ),

        "mean_absolute_difference":
            float(
                absolute_difference.mean()
            )
    }


all_predictions_match = bool(
    np.all(
        reproduced_predictions
        ==
        phase3_frozen_predictions_df[
            "learned_prediction"
        ].to_numpy()
    )
)


max_probability_difference = max(

    result[
        "max_absolute_difference"
    ]

    for result
    in probability_checks.values()
)


print("=" * 90)
print("PHASE 3 NUMERICAL REPRODUCTION")
print("=" * 90)

print(
    "\nAll predicted classes match:",
    all_predictions_match
)


for class_name, result in (
    probability_checks.items()
):

    print(
        f"\n{class_name}"
    )

    print(
        "  Max absolute probability difference:",
        f"{result['max_absolute_difference']:.12f}"
    )

    print(
        "  Mean absolute probability difference:",
        f"{result['mean_absolute_difference']:.12f}"
    )


print(
    "\nOverall maximum probability difference:",
    f"{max_probability_difference:.12f}"
)


assert all_predictions_match, (
    "Frozen classifier does not reproduce "
    "all Phase 3 predictions."
)


# CSV values may have undergone decimal serialization,
# so use a very small numerical tolerance rather than exact equality.

assert max_probability_difference < 1e-10, (
    "Probability reproduction discrepancy "
    "is larger than expected."
)


print(
    "\n✓ Frozen Phase 3 classifier reproduced successfully."
)
```

```python
# ================================================================
# PHASE 5B — CELL 25
# VERIFY FROZEN PHASE 2 QUANTITATIVE EQUATIONS
# ================================================================

def area_length_volume_ml(
    area_mm2,
    length_mm
):
    """
    Single-plane area-length volume.

    Inputs
    ------
    area_mm2 : float or array
        LV cavity area in mm^2.

    length_mm : float or array
        PCA major-axis length in mm.

    Returns
    -------
    volume_ml : float or array
        Estimated volume in mL.
    """

    area_cm2 = (
        np.asarray(
            area_mm2,
            dtype=float
        )
        / 100.0
    )

    length_cm = (
        np.asarray(
            length_mm,
            dtype=float
        )
        / 10.0
    )

    volume_ml = (
        8.0
        /
        (
            3.0
            *
            np.pi
        )
    ) * (
        area_cm2 ** 2
        /
        length_cm
    )

    return volume_ml


recomputed_EDV_ml = area_length_volume_ml(

    phase2_frozen_patient_df[
        "ED_PRED_area_mm2"
    ],

    phase2_frozen_patient_df[
        "ED_PRED_length_mm"
    ]
)


recomputed_ESV_ml = area_length_volume_ml(

    phase2_frozen_patient_df[
        "ES_PRED_area_mm2"
    ],

    phase2_frozen_patient_df[
        "ES_PRED_length_mm"
    ]
)


recomputed_SV_ml = (
    recomputed_EDV_ml
    -
    recomputed_ESV_ml
)


recomputed_EF_pct = (

    recomputed_SV_ml
    /
    recomputed_EDV_ml

) * 100.0


equation_checks = {

    "EDV_ml":
        np.max(
            np.abs(
                recomputed_EDV_ml
                -
                phase2_frozen_patient_df[
                    "ED_PRED_volume_ml"
                ].to_numpy()
            )
        ),

    "ESV_ml":
        np.max(
            np.abs(
                recomputed_ESV_ml
                -
                phase2_frozen_patient_df[
                    "ES_PRED_volume_ml"
                ].to_numpy()
            )
        ),

    "SV_ml":
        np.max(
            np.abs(
                recomputed_SV_ml
                -
                phase2_frozen_patient_df[
                    "PRED_SV_ml"
                ].to_numpy()
            )
        ),

    "EF_pct":
        np.max(
            np.abs(
                recomputed_EF_pct
                -
                phase2_frozen_patient_df[
                    "PRED_EF_pct"
                ].to_numpy()
            )
        )
}


print("=" * 100)
print("PHASE 2 EQUATION REPRODUCTION")
print("=" * 100)


for quantity, difference in (
    equation_checks.items()
):

    print(
        f"\n{quantity}"
    )

    print(
        "  Maximum absolute difference:",
        f"{difference:.12f}"
    )


overall_equation_difference = max(
    equation_checks.values()
)


assert overall_equation_difference < 1e-10


print(
    "\n✓ Area-length volume equation reproduced."
)

print(
    "✓ EDV/ESV → SV reproduced."
)

print(
    "✓ SV/EDV → EF reproduced."
)
```

```python
# ================================================================
# PHASE 5B — CELL 30
# EXACT FROZEN PREPROCESSING + PHYSICAL GEOMETRY
# Reconstructed directly from authoritative Phase 2A implementation
# ================================================================

import cv2
import nibabel as nib

from scipy.ndimage import binary_fill_holes


SEGMENTATION_RESOLUTION = 256
SEGMENTATION_THRESHOLD = 0.50


def load_nifti_2d(file_path):

    nii = nib.load(
        str(file_path)
    )

    data = np.asarray(
        nii.get_fdata()
    )

    if (
        data.ndim == 3
        and data.shape[-1] == 1
    ):
        data = data[:, :, 0]

    if data.ndim != 2:
        raise ValueError(
            f"Expected 2D NIfTI, got {data.shape}"
        )

    spacing = tuple(
        float(x)
        for x in nii.header.get_zooms()[:2]
    )

    return data, spacing


def preprocess_for_segmentation(
    image,
    target_size=256
):

    image = np.asarray(
        image,
        dtype=np.float32
    )

    original_h, original_w = image.shape

    resized = cv2.resize(
        image,
        (
            target_size,
            target_size
        ),
        interpolation=cv2.INTER_LINEAR
    )

    minimum = float(
        resized.min()
    )

    maximum = float(
        resized.max()
    )

    if maximum > minimum:

        normalized = (
            resized - minimum
        ) / (
            maximum - minimum
        )

    else:

        normalized = np.zeros_like(
            resized
        )

    scale_x = (
        original_w
        /
        target_size
    )

    scale_y = (
        original_h
        /
        target_size
    )

    return (
        normalized.astype(
            np.float32
        ),
        (
            scale_x,
            scale_y
        )
    )


def effective_spacing(
    native_spacing,
    scale_factors
):

    spacing_x, spacing_y = map(
        float,
        native_spacing
    )

    scale_x, scale_y = map(
        float,
        scale_factors
    )

    return (
        spacing_x * scale_x,
        spacing_y * scale_y
    )


def keep_largest_component(mask):

    mask_uint8 = (
        np.asarray(mask)
        .astype(np.uint8)
    )

    if mask_uint8.sum() == 0:

        return np.zeros_like(
            mask_uint8,
            dtype=bool
        )

    n_labels, labels, stats, _ = (
        cv2.connectedComponentsWithStats(
            mask_uint8,
            connectivity=8
        )
    )

    if n_labels <= 1:

        return np.zeros_like(
            mask_uint8,
            dtype=bool
        )

    largest_label = (
        1
        +
        np.argmax(
            stats[
                1:,
                cv2.CC_STAT_AREA
            ]
        )
    )

    return (
        labels
        ==
        largest_label
    )


def postprocess_prediction(
    probability_mask,
    threshold=0.50
):

    binary_mask = (
        probability_mask
        >=
        threshold
    )

    largest = keep_largest_component(
        binary_mask
    )

    filled = binary_fill_holes(
        largest
    )

    return np.asarray(
        filled
    ).astype(bool)


def mask_area_mm2(
    mask,
    spacing
):

    mask = np.asarray(
        mask
    ).astype(bool)

    spacing_x, spacing_y = map(
        float,
        spacing
    )

    return float(
        mask.sum()
        *
        spacing_x
        *
        spacing_y
    )


def mask_points_mm(
    mask,
    spacing
):

    mask = np.asarray(
        mask
    ).astype(bool)

    rows, cols = np.where(
        mask
    )

    spacing_x, spacing_y = map(
        float,
        spacing
    )

    # Exact Phase 2A coordinate convention:
    # x = image columns
    # y = image rows

    x_mm = (
        cols.astype(np.float64)
        *
        spacing_x
    )

    y_mm = (
        rows.astype(np.float64)
        *
        spacing_y
    )

    return np.column_stack([
        x_mm,
        y_mm
    ])


def pca_major_axis_mm(
    mask,
    spacing
):

    points = mask_points_mm(
        mask,
        spacing
    )

    if len(points) < 2:
        return np.nan

    centered = (
        points
        -
        points.mean(
            axis=0
        )
    )

    covariance = np.cov(
        centered,
        rowvar=False
    )

    eigenvalues, eigenvectors = (
        np.linalg.eigh(
            covariance
        )
    )

    principal_axis = (
        eigenvectors[
            :,
            np.argmax(
                eigenvalues
            )
        ]
    )

    projections = (
        centered
        @
        principal_axis
    )

    return float(
        projections.max()
        -
        projections.min()
    )


def frozen_area_length_volume_ml(
    area_mm2,
    length_mm
):

    if (
        not np.isfinite(area_mm2)
        or not np.isfinite(length_mm)
        or area_mm2 <= 0
        or length_mm <= 0
    ):
        return np.nan

    area_cm2 = (
        area_mm2
        /
        100.0
    )

    length_cm = (
        length_mm
        /
        10.0
    )

    volume_ml = (
        8.0
        /
        (
            3.0
            *
            np.pi
        )
        *
        (
            area_cm2 ** 2
        )
        /
        length_cm
    )

    return float(
        volume_ml
    )


print("=" * 80)
print("FROZEN PHASE 2 MEASUREMENT FUNCTIONS")
print("=" * 80)

print("\n✓ 256×256 stretch preprocessing")
print("✓ Post-resize per-image min-max normalization")
print("✓ Anisotropic effective spacing")
print("✓ Threshold >= 0.50")
print("✓ 8-connected largest component")
print("✓ Binary hole filling")
print("✓ Physical LV area")
print("✓ Exact PCA major-axis length")
print("✓ Single-plane area-length volume")
```

```python
# ================================================================
# PHASE 5B — CELL 31
# LOAD EXACT FROZEN PHASE 1 SEGMENTATION MODEL
# ================================================================

import gc
import tensorflow as tf


SMOOTH = 1e-6


def dice_coefficient(
    y_true,
    y_pred
):

    y_true = tf.cast(
        y_true,
        tf.float32
    )

    y_pred = tf.cast(
        y_pred,
        tf.float32
    )

    intersection = tf.reduce_sum(
        y_true * y_pred,
        axis=[1, 2, 3]
    )

    denominator = (
        tf.reduce_sum(
            y_true,
            axis=[1, 2, 3]
        )
        +
        tf.reduce_sum(
            y_pred,
            axis=[1, 2, 3]
        )
    )

    dice = (
        2.0 * intersection
        +
        SMOOTH
    ) / (
        denominator
        +
        SMOOTH
    )

    return tf.reduce_mean(
        dice
    )


def dice_loss(
    y_true,
    y_pred
):

    return (
        1.0
        -
        dice_coefficient(
            y_true,
            y_pred
        )
    )


def bce_dice_loss(
    y_true,
    y_pred
):

    bce = (
        tf.keras.losses
        .binary_crossentropy(
            y_true,
            y_pred
        )
    )

    bce = tf.reduce_mean(
        bce
    )

    return (
        bce
        +
        dice_loss(
            y_true,
            y_pred
        )
    )


CUSTOM_OBJECTS = {

    "bce_dice_loss":
        bce_dice_loss,

    "dice_loss":
        dice_loss,

    "dice_coefficient":
        dice_coefficient
}


gc.collect()


frozen_segmentation_model = (
    tf.keras.models.load_model(
        FROZEN_SEGMENTATION_MODEL_PATH,
        custom_objects=CUSTOM_OBJECTS,
        compile=False
    )
)


assert (
    frozen_segmentation_model.input_shape
    ==
    (None, 256, 256, 1)
)

assert (
    frozen_segmentation_model.output_shape
    ==
    (None, 256, 256, 1)
)

assert (
    frozen_segmentation_model.count_params()
    ==
    7_759_521
)


print("=" * 80)
print("FROZEN SEGMENTATION MODEL LOADED")
print("=" * 80)

print(
    "\nParameters:",
    f"{frozen_segmentation_model.count_params():,}"
)

print(
    "Input shape :",
    frozen_segmentation_model.input_shape
)

print(
    "Output shape:",
    frozen_segmentation_model.output_shape
)

print(
    "\n✓ Frozen Phase 1 model ready for inference."
)
```

```python
# ================================================================
# PHASE 5B — CELL 33
# SINGLE-FRAME FROZEN SEGMENTATION → MEASUREMENT FUNCTION
# ================================================================

def analyze_lv_frame(
    image,
    native_spacing
):

    # ------------------------------------------------------------
    # 1. Frozen preprocessing
    # ------------------------------------------------------------

    processed_image, scale_factors = (
        preprocess_for_segmentation(
            image,
            target_size=
                SEGMENTATION_RESOLUTION
        )
    )


    resized_spacing = effective_spacing(
        native_spacing,
        scale_factors
    )


    # ------------------------------------------------------------
    # 2. Frozen segmentation inference
    # ------------------------------------------------------------

    model_input = (
        processed_image[
            None,
            ...,
            None
        ]
    )


    probability_mask = (
        frozen_segmentation_model(
            model_input,
            training=False
        )
        .numpy()[0, :, :, 0]
    )


    predicted_mask = (
        postprocess_prediction(
            probability_mask,
            threshold=
                SEGMENTATION_THRESHOLD
        )
    )


    # ------------------------------------------------------------
    # 3. Frozen quantitative geometry
    # ------------------------------------------------------------

    area_mm2 = mask_area_mm2(
        predicted_mask,
        resized_spacing
    )


    length_mm = pca_major_axis_mm(
        predicted_mask,
        resized_spacing
    )


    volume_ml = (
        frozen_area_length_volume_ml(
            area_mm2,
            length_mm
        )
    )


    # ------------------------------------------------------------
    # 4. Structured output
    # ------------------------------------------------------------

    return {

        "processed_image":
            processed_image,

        "probability_mask":
            probability_mask,

        "predicted_mask":
            predicted_mask,

        "native_spacing":
            tuple(
                float(x)
                for x in native_spacing
            ),

        "effective_spacing":
            tuple(
                float(x)
                for x in resized_spacing
            ),

        "scale_factors":
            tuple(
                float(x)
                for x in scale_factors
            ),

        "area_mm2":
            float(area_mm2),

        "length_mm":
            float(length_mm),

        "volume_ml":
            float(volume_ml)
    }


print(
    "✓ Single-frame frozen inference function ready."
)
```

```python
# ================================================================
# PHASE 5B — CELL 38
# COMPLETE FROZEN PATIENT-LEVEL INFERENCE PIPELINE
# ================================================================

def analyze_a4c_pair(
    ed_image,
    es_image,
    ed_native_spacing,
    es_native_spacing
):
    """
    Complete frozen assistive A4C analysis pipeline.

    Required input
    --------------
    ed_image, es_image:
        Identified A4C ED and ES 2D frames.

    ed_native_spacing, es_native_spacing:
        Original physical pixel spacing from the source images.

    Returns
    -------
    Structured segmentation, quantitative, and EF-band outputs.

    Important
    ---------
    This is an assistive, non-diagnostic pipeline.
    The classifier confidence is an uncalibrated model
    confidence score and must not be interpreted as a
    patient-level reliability or safety guarantee.
    """

    # ============================================================
    # PHASE 1 + PHASE 2
    # ============================================================

    ed_result = analyze_lv_frame(
        ed_image,
        ed_native_spacing
    )

    es_result = analyze_lv_frame(
        es_image,
        es_native_spacing
    )


    # ------------------------------------------------------------
    # Quantitative patient-level measurements
    # ------------------------------------------------------------

    edv_ml = float(
        ed_result["volume_ml"]
    )

    esv_ml = float(
        es_result["volume_ml"]
    )


    if (
        not np.isfinite(edv_ml)
        or not np.isfinite(esv_ml)
        or edv_ml <= 0
    ):
        raise ValueError(
            "Valid EDV/ESV could not be derived "
            "from the predicted LV masks."
        )


    sv_ml = float(
        edv_ml
        -
        esv_ml
    )


    ef_pct = float(
        sv_ml
        /
        edv_ml
        *
        100.0
    )


    # ============================================================
    # PHASE 3 FEATURE VECTOR
    # Exact frozen feature order
    # ============================================================

    feature_values = np.array(
        [[
            ed_result["area_mm2"],
            es_result["area_mm2"],

            ed_result["length_mm"],
            es_result["length_mm"],

            edv_ml,
            esv_ml,

            sv_ml,
            ef_pct
        ]],
        dtype=float
    )


    assert (
        feature_values.shape
        ==
        (1, 8)
    )


    # ============================================================
    # PHASE 3 — FROZEN CLASSIFIER
    # ============================================================

    predicted_class = (
        frozen_phase3_classifier
        .predict(
            feature_values
        )[0]
    )


    class_probabilities = (
        frozen_phase3_classifier
        .predict_proba(
            feature_values
        )[0]
    )


    classifier_classes = (
        frozen_phase3_classifier
        .classes_
    )


    class_scores = {

        str(class_name):
            float(score)

        for class_name, score
        in zip(
            classifier_classes,
            class_probabilities
        )
    }


    model_confidence_score = float(
        np.max(
            class_probabilities
        )
    )


    # ============================================================
    # STRUCTURED APPLICATION OUTPUT
    # ============================================================

    return {

        "segmentation": {

            "ED": {
                "processed_image":
                    ed_result[
                        "processed_image"
                    ],

                "probability_mask":
                    ed_result[
                        "probability_mask"
                    ],

                "predicted_mask":
                    ed_result[
                        "predicted_mask"
                    ],

                "effective_spacing":
                    ed_result[
                        "effective_spacing"
                    ]
            },

            "ES": {
                "processed_image":
                    es_result[
                        "processed_image"
                    ],

                "probability_mask":
                    es_result[
                        "probability_mask"
                    ],

                "predicted_mask":
                    es_result[
                        "predicted_mask"
                    ],

                "effective_spacing":
                    es_result[
                        "effective_spacing"
                    ]
            }
        },


        "quantitative_measurements": {

            "ED_area_mm2":
                float(
                    ed_result[
                        "area_mm2"
                    ]
                ),

            "ES_area_mm2":
                float(
                    es_result[
                        "area_mm2"
                    ]
                ),

            "ED_length_mm":
                float(
                    ed_result[
                        "length_mm"
                    ]
                ),

            "ES_length_mm":
                float(
                    es_result[
                        "length_mm"
                    ]
                ),

            "EDV_ml":
                edv_ml,

            "ESV_ml":
                esv_ml,

            "SV_ml":
                sv_ml,

            "EF_pct":
                ef_pct
        },


        "classification": {

            "estimated_EF_band":
                str(
                    predicted_class
                ),

            "class_scores":
                class_scores,

            "model_confidence_score":
                model_confidence_score,

            "confidence_interpretation":
                (
                    "Uncalibrated model confidence score; "
                    "not a patient-level reliability or "
                    "safety guarantee."
                )
        },


        "system": {

            "view":
                "A4C",

            "input_requirement":
                (
                    "Paired identified ED and ES frames "
                    "with physical pixel-spacing metadata."
                ),

            "purpose":
                (
                    "Assistive LV analysis."
                ),

            "diagnostic_status":
                "Non-diagnostic"
        }
    }


print("=" * 85)
print("COMPLETE FROZEN A4C PIPELINE")
print("=" * 85)

print(
    "\nED/ES input"
    "\n   ↓"
    "\nFrozen LV segmentation"
    "\n   ↓"
    "\nPhysical LV geometry"
    "\n   ↓"
    "\nEDV / ESV / SV / EF"
    "\n   ↓"
    "\nFrozen EF-band classifier"
    "\n   ↓"
    "\nEF band + uncalibrated confidence"
)

print(
    "\n✓ analyze_a4c_pair() ready."
)
```

```python
# ================================================================
# PHASE 5B — CELL 40
# 50-PATIENT COMPLETE END-TO-END REPRODUCTION
# ================================================================

import gc
import time

end_to_end_records = []
start_time = time.time()

frozen_lookup = (
    phase3_frozen_predictions_df
    .set_index("patient_id")
)

assert len(frozen_lookup) == 50
assert frozen_lookup.index.is_unique

for index, patient_id in enumerate(
    frozen_lookup.index.tolist(), start=1
):
    patient_dir = DATASET_PATH / patient_id

    ed_image, ed_spacing = load_nifti_2d(
        patient_dir / f"{patient_id}_4CH_ED.nii.gz"
    )

    es_image, es_spacing = load_nifti_2d(
        patient_dir / f"{patient_id}_4CH_ES.nii.gz"
    )

    result = analyze_a4c_pair(
        ed_image,
        es_image,
        ed_spacing,
        es_spacing
    )

    predicted_band = result["classification"]["estimated_EF_band"]
    class_scores = result["classification"]["class_scores"]

    frozen_row = frozen_lookup.loc[patient_id]

    record = {
        "patient_id": patient_id,
        "frozen_prediction": frozen_row["learned_prediction"],
        "integrated_prediction": predicted_band,
        "prediction_match": (
            predicted_band == frozen_row["learned_prediction"]
        )
    }

    for class_name in frozen_phase3_classifier.classes_:
        record[f"frozen_prob_{class_name}"] = float(
            frozen_row[f"prob_{class_name}"]
        )
        record[f"integrated_prob_{class_name}"] = float(
            class_scores[class_name]
        )

    end_to_end_records.append(record)

    del result, ed_image, es_image

    if index % 10 == 0:
        gc.collect()
        print(f"Processed {index}/50 patients")

end_to_end_df = pd.DataFrame(end_to_end_records)

probability_difference_columns = []

for class_name in frozen_phase3_classifier.classes_:
    difference_column = f"abs_diff_{class_name}"

    end_to_end_df[difference_column] = (
        end_to_end_df[f"frozen_prob_{class_name}"]
        -
        end_to_end_df[f"integrated_prob_{class_name}"]
    ).abs()

    probability_difference_columns.append(difference_column)

prediction_matches = int(end_to_end_df["prediction_match"].sum())

maximum_probability_difference = float(
    end_to_end_df[probability_difference_columns]
    .to_numpy()
    .max()
)

probabilities_reproduced = bool(
    np.allclose(
        end_to_end_df[
            [f"frozen_prob_{c}" for c in frozen_phase3_classifier.classes_]
        ].to_numpy(),
        end_to_end_df[
            [f"integrated_prob_{c}" for c in frozen_phase3_classifier.classes_]
        ].to_numpy(),
        atol=1e-10,
        rtol=0
    )
)

print("\n" + "=" * 85)
print("PHASE 5B — COMPLETE 50-PATIENT VERIFICATION")
print("=" * 85)

print(f"Prediction matches: {prediction_matches}/50")
print(f"Maximum probability difference: {maximum_probability_difference:.12g}")
print(f"Probabilities reproduced within 1e-10: {probabilities_reproduced}")
print(f"Elapsed time: {time.time() - start_time:.1f} seconds")

display(
    end_to_end_df[
        ["patient_id", "frozen_prediction",
         "integrated_prediction", "prediction_match"]
    ].head(10)
)

if prediction_matches == 50 and probabilities_reproduced:
    print("\n✓ FULL END-TO-END REPRODUCTION PASSED")
else:
    print("\n⚠ Reproduction mismatch — investigate before freezing Phase 5B")
    display(end_to_end_df.loc[~end_to_end_df["prediction_match"]])
```
